```python
# Weather Finder

import requests

url = "https://geocoding-api.open-meteo.com/v1/search"

city = input("Enter your city: ").strip()

response = requests.get(url, params={"name": city})

data = response.json()

if "results" not in data or len(data["results"]) == 0:
    
    print("Error: City not found!")

else:
    
    city_data = data["results"][0]

    print("City:", city_data["name"])
    
    print("Country:", city_data["country"])

    latitude = city_data["latitude"]
    
    longitude = city_data["longitude"]

    weather_url = "https://api.open-meteo.com/v1/forecast"

    try:
        response = requests.get(weather_url, params={"latitude": latitude, "longitude": longitude, "current": "temperature_2m"},timeout=10)

        response.raise_for_status()
        
        weather_data = response.json()

        print("Temperature:", weather_data["current"]["temperature_2m"], "°C")

    except requests.exceptions.RequestException:
        
        print("Error: Could not get weather data!")
```
